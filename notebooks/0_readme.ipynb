# Companion notebooks for _Achieving practical runtime advantage with quantum Markov chains_

This series of notebooks serves to reproduce the results and plots for the paper [*Achieving practical runtime advantage with quantum Markov chains*](www.arxiv.com) and showcases the details of the numerical simulations and models.

**Table of contents**

* `1_sherrington_kirkpatrick.ipynb`: generation and loading of the Sherrington-Kirkpatrick instances used in the experiments; average-case upper bounds on the Hamiltonian 1-norm and operator norm; loading of the precomputed hyperparameters for the quantum-enhanced Monte Carlo trial move; computation and visualization of spectral gaps and classical query counts for the relevant transition matrices.
* `2_classical_models.ipynb`: overview of the classical hardware baselines used to convert query counts into runtime estimates; comparison of CPU, GPU, and FPGA execution models; summary of the latency and throughput assumptions used in the subsequent implementation notebooks.
* `2.1_fixed_point_arithmetic.ipynb`: fixed-point arithmetic model for the FPGA implementation; discretization-error analysis for normalized Hamiltonian coefficients and energy differences; polynomial approximation of the Metropolis acceptance function; cutoff construction for the exponential tail and resulting operation-error estimates.
* `2.2_implementation_cpu.ipynb`: CPU implementation and timing model for dense SK Metropolis updates; description of the C++ benchmark, compilation settings, and measured timings; interpolation of the per-query cost for local spin-flip and uniform moves.
* `2.3_implementation_gpu.ipynb`: GPU implementation and timing model for dense SK Metropolis updates; description of the CUDA/C++ kernels, memory layout, random-number pools, and benchmark configuration; throughput-oriented timing analysis for local and uniform moves on the NVIDIA H100 NVL GPU.
* `2.4_implementation_fpga.ipynb`: FPGA implementation and timing model for local and uniform Metropolis operations; description of the HLS kernels, fixed-point formats, Chebyshev acceptance circuit, testbenches, synthesis scripts, and Vitis HLS reports; interpolation and projection of FPGA latencies.

**Notation**

| Symbol | Meaning |
|---|---|
| $n$ | Number of spins. |
| $\beta$ | Inverse temperature. |
| $\pi$ | Gibbs distribution $e^{-\beta H(x)}$. |
| $T_{yx}$ | Entry of the column-stochastic trial-move matrix $T$, giving the probability of proposing state $y$ starting from current state $x$. |
| $A_{yx}$ | Acceptance probability associated with the proposed move $x\to y$, i.e. the probability of accepting $y$ after it has been sampled from the trial distribution $T_{\cdot x}$. |
| $P_{yx}$ | Entry of the column-stochastic transition matrix $P$, giving the total probability that the Markov chain moves from current state $x$ to next state $y$. |
