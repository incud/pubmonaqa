# Companion notebooks for _Achieving practical runtime advantage with quantum Markov chains_

This series of notebooks serves to reproduce the results and plots for the paper [*Achieving practical runtime advantage with quantum Markov chains*](www.arxiv.com) and showcases the details of the numerical simulations and models.

**Table of contents**

* `1_model_and_queries.ipynb`: prepares the SK instances used throughout the experiments and studies the spectral-gap and query-count behavior of the relevant Markov chains.
* `2_classical_models.ipynb`: introduces the classical runtime baselines and explains how query counts are converted into CPU, GPU, and FPGA time estimates.
* `2.1_fixed_point_arithmetic.ipynb`: develops the fixed-point arithmetic model used for the FPGA estimates, including discretization and acceptance-function approximation errors.
* `2.2_implementation_cpu.ipynb`: presents the CPU benchmark for dense SK Metropolis updates and the fitted per-query timing model.
* `2.3_implementation_gpu.ipynb`: presents the GPU benchmark for dense SK Metropolis updates and the corresponding throughput-based timing model.
* `2.4_implementation_fpga.ipynb`: presents the FPGA implementation model for Metropolis updates and the latency estimates obtained from HLS synthesis.
* `3_quantum_walks.ipynb`: construction of the Szegedy quantum-walk used in the quantum Markov-chain runtime model.
* `3.1_fully_phase_arithmetic.ipynb`: develops a phase-based Boltzmann coin for the Metropolis rule and studies its polynomial approximation cost.
* `3.2_hybrid_arithmetic.ipynb`: develops the hybrid Boltzmann coin used in the final resource estimates, combining reversible energy arithmetic with a phase-based exponential approximation.
* `3.3_validation_resources.ipynb`: checks that the symbolic resource formulas used in the analysis match the implemented Qiskit circuits.



**Notation**

| Symbol | Meaning |
|---|---|
| $n$ | Number of spins. |
| $\beta$ | Inverse temperature. |
| $\pi$ | Gibbs distribution $e^{-\beta H(x)}$. |
| $T_{yx}$ | Entry of the column-stochastic trial-move matrix $T$, giving the probability of proposing state $y$ starting from current state $x$. |
| $A_{yx}$ | Acceptance probability associated with the proposed move $x\to y$, i.e. the probability of accepting $y$ after it has been sampled from the trial distribution $T_{\cdot x}$. |
| $P_{yx}$ | Entry of the column-stochastic transition matrix $P$, giving the total probability that the Markov chain moves from current state $x$ to next state $y$. |
