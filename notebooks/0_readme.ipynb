# Companion notebooks for *Practical advantage beyond the quadratic speedup limit with fully quantum walks*

This collection of notebooks reproduces the numerical results, resource estimates, and figures presented in the paper [*Practical advantage beyond the quadratic speedup limit with fully quantum walks*](https://arxiv.org/abs/XXXX.XXXXX). It also documents the simulation methods, hardware models, and circuit-resource calculations underlying the analysis.

**Table of contents**

* `1_model_and_queries.ipynb`: generates the Sherrington-Kirkpatrick instances used throughout the study and analyzes the spectral gaps and query complexities of the corresponding Markov chains.
* `2_classical_models.ipynb`: introduces the classical runtime baselines and explains how classical query counts are converted into CPU, GPU, and FPGA runtime estimates.
* `2.1_fixed_point_arithmetic.ipynb`: develops the fixed-point representation used by the FPGA implementation, including the discretization error and the approximation of the Metropolis acceptance function.
* `2.2_implementation_cpu.ipynb`: presents the CPU benchmark for dense SK Metropolis updates and derives the fitted single-step latency models.
* `2.3_implementation_gpu.ipynb`: presents the GPU benchmark for dense SK Metropolis updates and derives the corresponding single-chain kernel-latency models.
* `2.4_implementation_fpga.ipynb`: describes the FPGA implementation of the Metropolis kernels and derives latency and resource estimates from HLS synthesis.
* `3_quantum_walks.ipynb`: introduces the Szegedy quantum-walk operator used in the quantum sampling algorithm and summarizes the logical resources of its main components.
* `3.1_fully_phase_arithmetic.ipynb`: develops the fully phase-based implementation of the Boltzmann coin and analyzes its polynomial-approximation and logical-resource costs.
* `3.2_hybrid_arithmetic.ipynb`: develops the hybrid Boltzmann coin used in the final resource estimates, combining reversible fixed-point energy arithmetic with phase-based approximation of the square-root exponential.
* `3.3_validation_resources.ipynb`: verifies that the analytical logical-qubit counts and non-Clifford-depth bounds agree with the implemented Qiskit circuits.
* `4_quantum_models.ipynb`: combines the spectral-filtering algorithm, heuristic annealing schedule, Zeno-rewind query analysis, logical circuit resources, Clifford+$\phi$ fault-tolerant model, error budget, surface-code model, and end-to-end quantum runtime estimate.
* `5_reproduce_plot.ipynb`: reproduces the figures in the main text and appendices from the processed numerical data and plotting utilities. The circuit diagrams are prepared separately using Quantikz.
