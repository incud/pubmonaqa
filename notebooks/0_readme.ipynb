# Companion notebooks for _Achieving practical runtime advantage with quantum Markov chains_

This series of notebooks serves to reproduce the results and plots for the paper [*Achieving practical runtime advantage with quantum Markov chains*](www.arxiv.com) and showcases the details of the numerical simulations and models.

**Table of contents**

* `1_sherrington_kirkpatrick.ipynb`: generation and loading of the Sherrington-Kirkpatrick instances used in the experiments; average-case upper bounds on the Hamiltonian 1-norm and operator norm; loading of the precomputed hyperparameters for the quantum-enhanced Monte Carlo trial move; computation and visualization of spectral gaps and classical query counts for the relevant transition matrices.
